```python
import math

inputs = [3, 12, 5, 7]

t = inputs[0]

for i in range(1, t + 1):
    n = inputs[i]

    if n <= 1:
        print("Not prime")
    elif n == 2:
        print("Prime")
    elif n % 2 == 0:
        print("Not prime")
    else:
        is_prime = True
        for j in range(3, int(math.sqrt(n)) + 1, 2):
            if n % j == 0:
                is_prime = False
                break

        print("Prime" if is_prime else "Not prime")
```
